```python
import numpy as np 
import matplotlib.pyplot as plt

#variavesde interessa
from sympy import Symbol
#Distribuitiva
from sympy import expand
#Calculo de diferencias
from sympy import diff
# Resolucao de sistemas lineares
from sympy import solve
```

```python
x_obs = [0,1,2,3,4]
y_obs = [1,3,2,4,5]

for i in range(len(x_obs)):
    print(f"{y_obs[i]} - m*{x_obs[i]} - b")

m = Symbol ("m")
b = Symbol ("b")

d_sqr = []

for i in range(len(x_obs)):
    expr = (y_obs[i] - m*x_obs[i] -b)
    print(expand(expr**2))
    d_sqr.append(expand(expr**2))


d_sum = 0
for i in range(len(d_sqr)):
    d_sum = expand(d_sum+d_sqr[i])
d_sum
```
